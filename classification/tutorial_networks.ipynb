networks.py에서 정의한 CustomNetwork을 불러옵니다. <br> <br>

train option에 있는 하이퍼파라미터를 CustomNetwork에 입력하여 초기화합니다. <br> <br>

```python
from networks import init_network

network = init_network(network)
```

networks.py에서 정의한 init_network 함수를 불러옵니다. <br> <br>
이 함수는 네트워크의 변수를 적절한 값으로 초기화해주는 역할을 수행합니다. <br> <br>

```python
import torch.nn as nn

criterion = nn.CrossEntropyLoss()
print(criterion)
```

사용할 손실함수인 CrossEntropy를 불러옵니다. <br> <br>

```python
import torch.optim as optim

optimizer = optim.Adam(network.parameters(), lr=opt.lr)
print(optimizer)
```

사용할 최적화 함수(optimizer)를 불러옵니다. <br> <br>
여기서는 optim.Adam을 사용할 예정입니다. <br> <br>
optim.Adam에는 최소 두 가지를 입력하여 초기화해줘야 합니다. <br> <br>
(1) 최적화할 변수들 (network가 가지고 있는 변수들) <br>
(2) learning rate <br>

```python
from networks import define_network, define_criterion, define_optimizer
```

networks.py 안에는 <br> <br>
(1) network를 정의하고 자동으로 초기화해주는 define_network <br>
(2) 손실함수를 자동으로 불러와주는 define_criterion <br>
(3) optimizer를 자동으로 불러와주는 define_optimizer <br> <br>
세 가지가 정의되어 있습니다. <br> <br>

```python
network = define_network(opt)
print(network)

criterion = define_criterion(opt)
print(criterion)

optimizer = define_optimizer(network, opt)
print(optimizer)
```
